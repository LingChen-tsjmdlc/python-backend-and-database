## 6.2 类与对象、__init__、self

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def introduce(self):
        print(f"我是{self.name}，考了{self.score}分")

s1 = Student("小明", 85)
s2 = Student("小红", 92)
s1.introduce()
s2.introduce()
```

## 6.3 实例属性 vs 类属性

```python
class Student:
    school = "深圳大学"

    def __init__(self, name):
        self.name = name

s1 = Student("小明")
s2 = Student("小红")
print(s1.school, s2.school, Student.school)
print(s1.name, s2.name)
```

## 6.4 封装：__x 藏起来，@property 留只读入口

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

acc = Account("小明", 1000)
print(acc.balance)
```

## 6.5 继承、重写与 super()

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            print("余额不足")
            return
        self.balance -= amount

class CreditAccount(Account):
    def withdraw(self, amount):
        if amount > self.balance + 5000:
            print("超出透支额度")
            return
        self.balance -= amount

acc = CreditAccount("小明", 1000)
acc.withdraw(3000)
print(acc.balance)
```

## 6.6 多态：同一句话，各自理解

```python
class Basket:
    def __init__(self):
        self.items = []

    def put(self, item):
        self.items.append(item)

    def __len__(self):
        return len(self.items)

class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

b = Basket()
b.put("苹果")
b.put("香蕉")
t = Team(["小明", "小红", "小刚"])
print(len(b))
print(len(t))

class Dog:
    def speak(self):
        print("汪汪")

class Cat:
    def speak(self):
        print("喵喵")

for a in [Dog(), Cat(), Dog()]:
    a.speak()
```

## 6.7 魔术方法

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f"账户({self.owner}，余额{self.balance}元)"

    def __eq__(self, other):
        return self.owner == other.owner

print(Account("小明", 1000))
print(Account("小明", 1) == Account("小明", 9999))
```

## 6.8 实战：银行账户系统（完整版见 07_bank_account.py）

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            print("余额不足")
            return
        self._balance -= amount

    def __str__(self):
        return f"账户({self.owner}，余额{self._balance}元)"

acc = Account("小明", 1000)
acc.deposit(500)
acc.withdraw(2000)
print(acc)
```
