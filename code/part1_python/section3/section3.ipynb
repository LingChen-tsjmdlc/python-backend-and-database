## 3.1 条件判断 if / elif / else、match-case

条件成立才执行；多分支从上到下第一个成立就执行。

```python
score = 85
if score >= 90:
    print("优秀")
elif score >= 60:
    print("及格")
else:
    print("不及格")
```

```python
choice = "2"
match choice:
    case "1":
        print("添加")
    case "2":
        print("查看")
    case "3":
        print("删除")
    case _:
        print("输入无效")
```

## 3.2 for 循环：让代码重复执行

循环的本质是**重复执行一段逻辑**，遍历只是其中一种用法；`range` 是「循环固定次数」的补充工具（含头不含尾）。

```python
for ch in "abc":
    print(ch)

for i in range(5):
    print(i)

total = 0
for i in range(1, 101):
    total += i
print("1到100的和：", total)
```

### 嵌套循环

外层每转一圈，内层完整转完一轮（像钟表：分针转一圈，时针走一格）。

```python
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
```

## 3.3 while 循环

不知道循环几次、按条件停就用 while。注意更新条件变量，避免死循环。

```python
count = 1
while count <= 5:
    print(f"第 {count} 次")
    count += 1
```

## 3.4 break 与 continue

`break` 结束整个循环；`continue` 跳过本次。

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)

for i in range(1, 10):
    if i % 2 == 0:
        continue
    print(i)
```

## 3.5 实战：猜数字

先演示一轮判断；完整循环版（`while True` + `input()` + `random.randint`）见 05_guess_number.py。

```python
secret = 42          # 实际用 random.randint(1, 100) 随机生成
guess = 50           # 实际用 int(input(...)) 接收

if guess > secret:
    print("大了，再猜")
elif guess < secret:
    print("小了，再猜")
else:
    print("恭喜你，猜对了！")
```

## 3.5 实战：九九乘法表与打印图形

嵌套循环：外层管行、内层管列。

```python
for i in range(1, 10):
    for j in range(1, i + 1):
        print(f"{j}×{i}={i * j}", end="\t")
    print()
```

```python
for i in range(1, 6):
    print("*" * i)
```
