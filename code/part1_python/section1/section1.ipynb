## 3. 缩进（概念）

Python 用缩进（行首空格）划分代码块，同一层级要对齐，约定每级 4 个空格。

> 下面的 `if` 只是用来演示「缩进划分代码块」，`if` 的用法后面章节才讲，本节不要求掌握。

```python
if 10 > 5:
    print("10 比 5 大")
    print("这两行往里缩，属于 if")
print("这行没缩进，不属于 if")
```
