## 4.2 遍历与列表推导式（含 map/filter 了解）

```python
scores = [85, 92, 78, 90]
for i, s in enumerate(scores):
    print(f"第 {i + 1} 个：{s}")

names = ["小明", "小红", "小刚"]
for name, s in zip(names, scores):
    print(f"{name}：{s} 分")

doubled = [s * 2 for s in scores]
high = [s for s in scores if s >= 90]
print(doubled, high)
```

### map 与 filter

第一个参数传内置函数（如 `int`、`str`）就能用；返回的是迭代器，要 `list()` 才能看到结果。

```python
print(list(map(str, [1, 2, 3])))       # 每个元素过一遍 str()
print(list(map(int, ["85", "92"])))   # 批量字符串转数字
print(list(filter(None, ["85", "", "92"])))  # 滤掉空字符串
```

## 4.3 字典：键值对与 get

按键存取；键可能不存在时用 `get()` 更安全。

遍历用 `.items()`：它交出所有键值对，`for` 写两个名字分别接住键和值。

```python
student = {"name": "小明", "age": 18}
student["city"] = "深圳"
student["age"] = 19

print(student["name"])
print(student.get("phone"))
print(student.get("phone", "未知"))

for k, v in student.items():
    print(f"{k}: {v}")
```

## 4.3 字典推导式

```python
scores = {"小明": 85, "小红": 92, "小刚": 78}
passed = {n: s for n, s in scores.items() if s >= 80}
print(passed)
```

## 4.5 引用机制：为什么改一个，另一个也变了

`b = a` 不复制，两个名字共用同一个列表对象（可用 `id()` 验证）。
数字等不可变类型没这个问题——「改」其实是新建对象。

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
print(id(a) == id(b))   # True：同一个对象

a = 100
b = a
b = b + 1
print(a)                  # 100：数字不受影响
```

## 4.5 浅拷贝与深拷贝

浅拷贝只独立外层，内层仍共享；深拷贝完全独立。

```python
import copy

a = [[1, 2], [3, 4]]
b = a.copy()
b[0].append(99)
print("浅拷贝后 a:", a)

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0].append(99)
print("深拷贝后 a:", a)
print("深拷贝后 b:", b)
```

## 4.6 实战：带菜单的简易通讯录

下面是完整的交互版（与 06_contacts.py 相同）。**运行此 cell 会弹出输入框**，依次输入操作编号即可体验：添加 → 查看 → 查找 → 瞎输一个看报错 → 选 4 退出。

```python
"""实战：带菜单的简易通讯录，对应笔记 4.6。

通讯录 = 列表，里面每个元素是一个联系人的字典。
while True 循环显示菜单，直到选择退出。
"""

contacts = []

while True:
    print("\n===== 通讯录 =====")
    print("1. 添加联系人")
    print("2. 查看全部")
    print("3. 按姓名查找")
    print("4. 退出")
    choice = input("请选择操作（1-4）：")

    if choice == "1":
        name = input("请输入姓名：")
        phone = input("请输入电话：")
        contacts.append({"name": name, "phone": phone})
        print(f"已添加：{name}")

    elif choice == "2":
        if len(contacts) == 0:
            print("通讯录是空的")
        else:
            for c in contacts:
                print(f"{c['name']}：{c['phone']}")

    elif choice == "3":
        target = input("请输入要查找的姓名：")
        found = False
        for c in contacts:
            if c["name"] == target:
                print(f"找到了：{c['name']} 的电话是 {c['phone']}")
                found = True
        if not found:
            print(f"通讯录里没有 {target}")

    elif choice == "4":
        print("再见！")
        break

    else:
        print("输入无效，请输入 1-4")
```
