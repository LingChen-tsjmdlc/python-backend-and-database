## 5.1 为什么要用函数：写一遍，用无数次

```python
def report(class_name, scores):
    avg = sum(scores) / len(scores)
    print(f"{class_name}的平均分是 {avg:.1f}")

report("一班", [85, 92, 78])
report("二班", [60, 75, 88])
report("三班", [90, 91, 95])
```

## 5.2 参数：位置 / 默认 / 关键字

```python
def introduce(name, city="深圳"):
    print(f"我是{name}，来自{city}")

introduce("小明", "广州")
introduce("小明")
introduce(city="汕头", name="小刚")
```

## 5.2 参数：*args 与 **kwargs（个数不确定时）

```python
def average(*scores):
    print("scores 是元组：", scores)
    return sum(scores) / len(scores)

print(average(85, 92, 78))

def show_info(**info):
    for k, v in info.items():
        print(f"{k}: {v}")

show_info(name="小明", age=18)
```

## 5.3 返回值与作用域

```python
def average(scores):
    return sum(scores) / len(scores)

avg = average([85, 92, 78])
print(avg + 5)

def min_max(scores):
    return min(scores), max(scores)

lo, hi = min_max([85, 92, 78])
print(lo, hi)
```

## 5.3 闭包：函数带走它出生时的环境

```python
def make_adder(n):
    def add(x):
        return x + n
    return add

add5 = make_adder(5)
print(add5(1))
print(add5(10))
```

## 5.4 lambda 与高阶函数

```python
students = [
    {"name": "小明", "score": 85},
    {"name": "小红", "score": 92},
    {"name": "小刚", "score": 78},
]
ranked = sorted(students, key=lambda s: s["score"], reverse=True)
for s in ranked:
    print(s["name"], s["score"])
```

## 5.4 重新认识 map / filter

```python
scores = [85, 92, 78, 90]
print(list(map(lambda s: s / 10, scores)))
print(list(filter(lambda s: s >= 80, scores)))
```

## 5.5 常用内置函数；sorted(列表) vs 列表.sort()

```python
nums = [3, 1, 2]
a = sorted(nums)
print(a, nums)
b = nums.sort()
print(b, nums)
print(abs(-3), round(3.14159, 2), list(range(3)))
```

## 5.5 递归：函数调用自己

```python
def countdown(n):
    if n <= 0:
        print("点火！")
        return
    print(n)
    countdown(n - 1)

countdown(3)
```
