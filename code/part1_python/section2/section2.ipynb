## 2.1 变量

变量是带名字的存储位置，用 `=` 赋值；一处存、多处用；值可被覆盖。

```python
name = "小明"
print("欢迎你，"+ name)
print(name + "，你有 3 条新消息")

age = 18
age = 20   # 新值覆盖旧值
print(age)
```

## 2.2 标识符与命名规范

命名用 snake_case、见名知意、不能以数字开头、不能用关键字。

```python
user_name = "ok"
total_price = 99.5
age2 = 18
print(user_name, total_price, age2)
```

## 2.3 基本数据类型

四种基本类型 `int` / `float` / `str` / `bool`，用 `type()` 查看。

```python
age = 18
height = 1.75
name = "小明"
is_student = True

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

### 特殊值 None 与 NaN

`None` 表示「空 / 没有值」（类型 NoneType）；`NaN` 表示「非数字」（类型仍是 float）。

注意坑：`NaN` 不等于任何值，连自己都不等，判断要用 `math.isnan()`。

```python
import math

answer = None
print(answer, type(answer), answer is None)

x = float("nan")
print(x, type(x))
print(x == x)          # False
print(math.isnan(x))   # True
```

## 2.4 类型转换

`int()` / `float()` / `str()`；内容不合法会报错（如 `int("abc")`）。

```python
age = int("18")
print(age + 1)

height = float("1.75")
print(height)

print("成绩：" + str(95))
```

## 2.5 运算符

算术 / 比较 / 逻辑 / 复合赋值 / 成员与身份。

```python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)   # 算术
print(3 == 2, 3 != 2, 2 >= 2)          # 比较
print(3 > 2 and 5 > 1, not 3 > 2)      # 逻辑

count = 10
count += 5
print(count)                           # 复合赋值

print("a" in "abc")                    # 成员
print(None is None)                    # 身份
```

## 2.6 字符串

拼接、索引与切片（从 0 开始）、常用方法、f-string、转义字符。

```python
s = "Python"
print(s[0], s[-1])       # 索引
print(s[0:3], s[2:])     # 切片

print("abc".upper(), " hi ".strip(), len("abc"))   # 方法

name, age = "小明", 18
print(f"我叫{name}，今年{age}岁")   # f-string

print("第一行\n第二行")   # 转义
```

## 2.7 输入与输出

`input()` 拿到的永远是字符串，做数字运算要先转换。

> 注意：Notebook 里运行 `input()` 会弹出输入框、暂停等待。下面用固定值代替演示；实际脚本里可写 `age = int(input("请输入年龄："))`。

```python
# 演示：假设用户输入了 "18"
age_text = "18"          # 相当于 input() 拿到的字符串
age = int(age_text)
print(f"明年你 {age + 1} 岁")

# print 的格式化参数
print("2026", "08", "14", sep="-")
```

## 2.8 实战：BMI 计算器

BMI = 体重(kg) / 身高(m) 的平方。

> 同样为方便演示，这里用固定的身高体重；实际可用 `float(input(...))` 接收输入。

```python
height = 1.75   # 实际可用 float(input("请输入身高（米）："))
weight = 65     # 实际可用 float(input("请输入体重（公斤）："))

bmi = round(weight / (height ** 2), 1)
print(f"你的 BMI 是 {bmi}")
```
